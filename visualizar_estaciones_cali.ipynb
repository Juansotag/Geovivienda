# Visualización de Estaciones MIO (Cali)
Este notebook genera un mapa interactivo para visualizar las estaciones del MIO que están guardadas en el archivo local `estaciones_mio.geojson`.

```python
import folium
import json
import os

# Ruta al archivo GeoJSON
geojson_path = os.path.join('static', 'geo', 'cali', 'estaciones_mio.geojson')

if os.path.exists(geojson_path):
    with open(geojson_path, 'r', encoding='utf-8') as f:
        data = json.load(f)
    print(f"Cargadas {len(data.get('features', []))} estaciones.")
else:
    print("El archivo no existe. Asegúrate de haberlo guardado correctamente.")
    data = {'features': []}

# Crear el mapa centrado en Cali
m = folium.Map(location=[3.42, -76.52], zoom_start=12, tiles='CartoDB positron')

# Añadir marcadores para cada estación
for feat in data.get('features', []):
    coords = feat['geometry']['coordinates']
    nombre = feat['properties'].get('nombre', 'Estación MIO')
    folium.Marker(
        location=[coords[1], coords[0]],
        popup=nombre,
        icon=folium.Icon(color='red', icon='bus', prefix='fa')
    ).add_to(m)

# Mostrar el mapa en el notebook
m
```
